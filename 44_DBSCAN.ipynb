```python
from sklearn.cluster import DBSCAN 
import numpy as np
import matplotlib.pyplot as plt

X = np.array([[1,2],[2,2],[2,3],[8,7],[25,80]])
```

```python
plt.scatter(X[:,0],X[:,1])

plt.show()
```

```python
db = DBSCAN(eps = 10, min_samples = 3)

db.fit(X)
db.labels_
```

```python
from sklearn.datasets import make_circles
```

```python
# Concentric circles dataset
X, y = make_circles(n_samples = 500, factor = 0.6, noise = 0.03, random_state = 4)

# Apply DBSCAN tothe dataset
dbscan = DBSCAN(eps = 0.1, min_samples = 5)
Clusters = dbscan.fit_predict(X)

#plotting

plt.scatter(X[:,0], X[:,1], c = Clusters, cmap ='viridis', marker = 'o')
plt.title("DBSCAN Clustering of concentric Circles")
plt.xlabel("Features 0")
plt.ylabel("Features 1")
plt.show()
```
